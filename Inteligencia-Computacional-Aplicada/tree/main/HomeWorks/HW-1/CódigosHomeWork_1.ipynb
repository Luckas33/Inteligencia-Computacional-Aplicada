```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('/kaggle/input/student-career-success-prediction-dataset/student_career_success_dataset.csv')
```

```python
display(df.head())
```

```python
sns.histplot(data=df, x='Study_Hours_Per_Week', kde=True, color='blue')
plt.title('Estudo por semana') # Adiciona título
plt.show()
```

```python
sns.boxplot(data=df, x='Study_Hours_Per_Week', palette='Set1')
plt.title('Boxplot de Salário')
plt.show()
```

```python
sns.scatterplot(data=df, x='Study_Hours_Per_Week', y='Starting_Salary_USD')
plt.title('Idade vs Salário')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Solução 2: Jitter e Boxplot Sobreposto (RECOMENDADA)

# 1. Cria o Boxplot sutil para estatística, sem mostrar os outliers (showfliers=False)
sns.boxplot(
    data=df,
    x='Study_Hours_Per_Week',
    y='Starting_Salary_USD',
    palette='pastel',       # Cores pastéis sutis
    saturation=0.5,         # Menor saturação
    showfliers=False       # Oculta outliers do boxplot (pois já estão no stripplot)
)

# 2. Sobrepõe os pontos individuais com jitter e transparência
sns.stripplot(
    data=df,
    x='Projects_Completed',
    y='Starting_Salary_USD',
    color='darkblue',      # Cor escura contrastante
    size=3,                 # Pontos menores
    alpha=0.3,              # Transparência
    jitter=True            # Dispersão automática
)

# Formatação
plt.title('Estudo semanal vs Salário inicial (Solução para Sobreposição)')
plt.show()
```
